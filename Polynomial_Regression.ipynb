```python
# EDA 

plt.scatter(df["Temperature (°C)"],df["Ice Cream Sales (units)"])
plt.title("Plot the data")
plt.xlabel("Temperature (°C)")
plt.ylabel("Ice Cream Sales (units)")
plt.show()
```

```python
# Divide the data into x and y 

X = df["Temperature (°C)"].values.reshape(-1,1)
y = df["Ice Cream Sales (units)"]
```

```python
# polynomial regression

# Feature scaling

from sklearn.preprocessing import PolynomialFeatures

# create the instance

poly = PolynomialFeatures(degree=2)

# fit and transform

X_poly = poly.fit_transform(X)
```

```python
X_poly
```

```python
# Model building

from sklearn.linear_model import LinearRegression

# create the instance of model

LR = LinearRegression()

# Train the model on the X_poly

model = LR.fit(X_poly,y)

# model prediction
y_pred = model.predict(X_poly)
```

```python
# model evalution netrics

from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score

# find the MSE

MSE = mean_squared_error(y,y_pred)

# find the MAE
MAE = mean_absolute_error(y,y_pred)

# find the r2 score
r2_score = r2_score(y,y_pred)

# print the all metrics

print(f"MSE :{MSE}")
print(f"MAE :{MAE}")
print(f"r2_score :{r2_score}")
```
