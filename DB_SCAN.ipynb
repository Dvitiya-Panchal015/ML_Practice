```python
df.drop("true_group", axis = 1, inplace = True)
```

```python
# scatter plot 

plt.scatter(df["x"], df["y"])
plt.show()
```

```python
# import the models

from sklearn.cluster import DBSCAN

# create the instance of model 

model = DBSCAN(eps=2, min_samples=10)

# train the model
model.fit(df)

# display the labels
label=model.labels_
label

```

```python
# make the plot using the label

plt.scatter(df["x"],df["y"], c=label)
plt.title("DBSCAN")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```
